### Random Variables

A Random Variable is a variable that takes numerical values based on the outcome of a random experiment.

In simple words —

A random variable assigns a value to each possible outcome of a random event.

Example : 

Tossing a coin → Outcome: {Head, Tail}

Random Variable (X):  X = 1 if Head,   X = 0 if Tail

Rolling a die → Outcome: {1, 2, 3, 4, 5, 6}

Random Variable (X):  X = number that appears on the die

Types of Random Variables

Random variables are classified into two main types

##### Discrete Random Variable:
A Discrete Random Variable takes a finite or countable number of distinct values.

Examples:

Number of students present in a class

Number of heads in 5 coin tosses

##### Continuous Random Variable:
A Continuous Random Variable can take any value within a given range or interval (infinite possible values).

Examples:

Height or Weight of a person

Temperature, Time, Distance, Pressure

##### Probability Distributions:
A Probability Distribution describes how the probabilities are distributed among all possible values of a random variable.

In other words —

It shows how likely each outcome of a random variable is.

##### Types of Probability Distributions:
Probability distributions are mainly divided into two types based on the type of random variable

##### 1. Discrete Probability Distribution

A Discrete Probability Distribution gives the probability of each possible value of a discrete random variable.

Example:

##### Binomial Distribution:
The Binomial Distribution gives the probability of getting a fixed number of successes (k) in a fixed number of independent trials (n), where each trial has two possible outcomes (success or failure).

###### Insert Image for formula 

Where:
n = number of trials

k = number of successes

p = probability of success

(1−p) = probability of failure

##### 2.Continuous Probability Distribution:
It is a probability distribution of the continuous random variable

The probability is given by the area under the pdf on a specified range. Thus, the value of the probability density function is 0 at a particular point

The area under the curve is always equal to 1. Therefore, is considered as a probability distribution

“In a continuous distribution, probability is represented by area under the curve.
If we take a rectangle and keep reducing its width, the area keeps decreasing.
When the width shrinks to zero (a single point), the area becomes zero.
Hence, the probability at an exact point is zero.”

Example: 

##### Normal Distribution

Where:

μ: Mean (center of the curve)

σ: Standard Deviation (spread of data)

##### Q) The mean salaries of Data Scientists working in Hyderabad, India is calculated to be 7,00,000 INR with a standard deviation of 90,000 INR. The random variable salary of Data Scientists follows a normal distribution. 

a) What is the probability that a Data Scientist in Hyderabad has a salary more than 10,00,000 INR?

b) What is the probability that a Data Scientist in Hyderabad has a salary between 6,00,000 & 9,00,000 INR?

c) What is the probability that a Data Scientist in Hyderabad has a salary less than 3,00,000 INR?

```python
# a) 
mu    =  700000
sigma =  90000
z=(1000000-mu) / sigma 

P = 1 - stats.norm.cdf(z)
print(' a) Probability that a Data Scientist in Chennai has a salary more than 10,00,000 is %1.4f', P)
```

```python
#b) 

z1= (600000-mu) / sigma
z2= (900000-mu) / sigma

P = stats.norm.cdf(z2) - stats.norm.cdf(z1)
print(' b) probability that a Data Scientist in Chennai has a salary between 6,00,000 & 9,00,000 INR is %1.4f', P)
```

```python
#c) 

z = (300000-mu) / sigma 

P = stats.norm.cdf(z)
print(' c) Probability that a Data Scientist in Chennai has a salary less than 4,00,000 INR is %1.4f',P)
```

### Central Limit Theorem

Parameter:
The population quantities are termed as a parameter,In inferential statistics, sample is used to estimate the population parameters

Example: Population mean

Statistic:
Statistical measure computed from sample  observations, Also known as an estimator

Example: Sample mean



##### The Central Limit Theorem (CLT) states that:

When we take many random samples of a given size from any population (with any shape of distribution),the sampling distribution of the sample mean will tend to become approximately normal as the sample size increases — regardless of the population’s original distribution.

###### Sampling distribution: 
The distribution of the means of the samples is called as Sampling Dsitribution.

Even if the population is not normally distributed (e.g., skewed, uniform, etc.),
the distribution of sample means will approach a normal distribution as the sample size (n) becomes large — typically  n≥30.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# Step 1: Create a non-normal population (Exponential Distribution)

population = beta.rvs(a=2, b=5, size=100000)

# Step 2: Draw multiple samples and store their means

sample_size = 50
sample_means = []
for i in range(1000):
    sample = np.random.choice(population, size=sample_size)  # sample size = 50
    sample_means.append(np.mean(sample))

# Step 3: Plot the population and sample means
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(population, bins=50)
plt.title('Original Population (Exponential)')

plt.subplot(1,2,2)
plt.hist(sample_means, bins=30, color='orange')
plt.title('Distribution of Sample Means (Approx. Normal)')

plt.show()

```

### Theory of Estimation

The theory of estimation is a branch of statistics that deals with estimating unknown population parameters (like mean, proportion, or variance) using sample data.
Since it’s often impractical to study the entire population, we draw samples and use them to make inferences about the population.

#### Point Estimation

A point estimate is a single value calculated from sample data that is used to estimate an unknown population parameter (like mean or proportion).

Example: Using the sample mean (x̄) to estimate the population mean (μ).

#### Interval Estimation (Confidence Intervals)

Interval estimation gives a range of values (instead of a single value) within which the true population parameter is expected to lie, along with a level of confidence.


```python
### Z_Score Values 
print('z_alpha_by_2 values of 90% Confidence Interval',
      round(stats.norm.isf(0.1/2),2))
print('z_alpha_by_2 values of 95% Confidence Interval',
      round(stats.norm.isf(0.05/2),2))
print('z_alpha_by_2 values of 99% Confidence Interval',
      round(stats.norm.isf(0.01/2),2))
```

#### Confidence Interval (Population Standard Deviation Known)

When the population standard deviation (σ) is known, the Z-distribution is used to construct the confidence interval for the population mean (μ)

##### Q) A paediatrician wants to check the amount of sugar in the 100g pack of baby food produced by KidsGrow company. The medical journal states that a standard deviation of sugar in 100g pack is 8g. The paediatrician collects 37 packets of baby food and found that the average sugar is 24g. Find the 90% confidence interval for the population mean.

As the sample size is large (> 30), we use the Z-distribution to calculate the confidence interval.

```python
x_bar = 24
sigma = 8
n = 37
z_alpha_by_2 = 1.64
LL = x_bar - z_alpha_by_2 * (sigma/np.sqrt(n))
UL = x_bar + z_alpha_by_2 * (sigma/np.sqrt(n))
print(LL,UL)
```

```python
# number of observations in sample
n = 37

# given sample mean
x_bar = 24 

# given population standard deviation
sigma = 8

# calculate the 90% confidence interval
# pass the sample mean to the parameter, 'loc'
# pass the scaling factor (std_pop / n^(1/2)) to the parameter, 'scale'
interval = stats.norm.interval(0.90, loc = x_bar, scale = sigma / np.sqrt(n))

# print the 90% confidence interval
print('The 90% confidence interval of population mean is:', interval)
```

### Hypothesis Testing

Hypothesis testing is a statistical method used to make decisions or draw conclusions about a population by using sample data.

We start by making an assumption about the population (called a hypothesis) and then test whether the sample data supports or rejects that assumption.

Hypothesis testing helps us decide whether the result we see in a sample is real or just happened by chance.

### Terminologies

##### Null Hypothesis (H₀):

The Null Hypothesis is the statement that there is no difference, no effect, or no change.
It represents the existing or assumed situation.

H₀: The average height of students = 165 cm

##### Alternate Hypothesis (H₁ or Hₐ):

The Alternate Hypothesis is the opposite of the null.
It states that there is a difference, an effect, or a change.

H₁: The average height of students ≠ 165 cm


##### Two Tail test 

##### One Tail test


##### Left Tail test 


##### Right Tail Lest 


#### P-Value 

Calculate the p-value using your test statistic.

Compare the p-value with the level of significance (α).

If p ≤ α, reject H₀.

If p > α, do not reject H₀.

### Errors in Hypothesis Testing

Sometimes our decision can be wrong. There are two possible errors:

| Type of Error         | What Happens                          | Example                                              | Result             |
| --------------------- | ------------------------------------- | ---------------------------------------------------- | ------------------ |
| **Type I Error (α)**  | Rejecting a true null hypothesis      | Saying a medicine works when it actually doesn’t     | **False Positive** |
| **Type II Error (β)** | Not rejecting a false null hypothesis | Saying a medicine doesn’t work when it actually does | **False Negative** |


In short:

Type I Error: Wrongly rejecting a true H₀.

Type II Error: Failing to reject a false H₀.

##### Q.It is believed that average math score of female students is minimum 75 out of 100. A researcher survey claims that in a math test female students tend to score less marks than the average marks of 75 out of 100. Consider a sample of 24 female students and perform a hypothesis test to check the claim with 90% confidence.

```python
# read the students performance data 
df = pd.read_csv('mathscore_1ttest.csv')
```

```python
# Step 1: Decide the test to perform. 
# 1 sample t test

# Step 2: Hypothesis Construction. 
# H0 : mu >= 75
# H1 : mu < 75

# Step 3 : Decide the Direction of the test  
# Left tail test

# Step 4: Calculate test statistic & p value
# In the question it is given the class average is 75 so population mean is 75
# a = data,  alternative = direction of the test
test_stat,p_val = stats.ttest_1samp(a=df[df['gender']=='female']['math score'],popmean=75,alternative='less')

print('test_statistic_value',test_stat)
print('p_value',p_val)

# Step 5: Conclusion
sig_level = 0.1
print('sig_level',sig_level)
if p_val < sig_level:
    print('Since p_val is less than sig_level We reject the Null Hypothesis')
else:
    print('Since p_val is not less than sig_level We fail to reject the Null Hypothesis')
```

#### Manual Calculation for reference using Formula

1_samp_t_test = (x_bar-mu)/(S/np.sqrt(n))

```python
x_bar = df[df['gender']=='female']['math score'].mean()
S = df[df['gender']=='female']['math score'].std()
mu = 75
n = 24
test_stat = (x_bar-mu)/(S/np.sqrt(n))
print('test_stat',test_stat)
cv = stats.t.isf(0.1,df=n-1)
print('critical value',-cv)
p_value = stats.t.cdf(test_stat,n-1)
print('p_value',p_value)
sig_level = 0.1
print('sig_level',sig_level)
p_value < sig_level
```

##### Q.It is believed that average writing score of male students is maximum 85 out of 100. A researcher survey claims that in writing test male students tend to score high marks than the average marks of 85  out of 100. ### 2 sample t-test

##### Q. The teachers' association claims that the total score of the students who completed the test preparation course is different than the total score of the students who have not completed the course. The collge management claim the average scores of both the groups is same. The sample data consists of 15 students who completed the course and 18 students who have not completed the course. Test the association's claim with ⍺ = 0.05.

```python
# read the students performance data 
df = pd.read_csv('totalmarks_2ttest.csv')
df['test preparation course'].unique()
```

```python

# Step 1: Decide the test to perform. 
# 2 sample t test

# Step 2: Hypothesis Construction. 
# H0 : mu1 = mu2
# H1 : mu1 != mu2

# Step 3 : Decide the Direction of the test  
# two tail test

# Step 4: Calculate test statistic & p value
course_completed = df[df['test preparation course']=='completed']['total score']
course_not_completed = df[df['test preparation course']=='none']['total score']
test_stat,p_val = stats.ttest_ind(course_completed,course_not_completed)
print('test_statistic_value',test_stat)
print('p_value',p_val)

# Step 5: Conclusion 
sig_level = 0.05
print('sig_level',sig_level)
if p_val < sig_level:
    print('Since p_val is less than sig_level We reject the Null Hypothesis')
else:
    print('Since p_val is not less than sig_level We fail to reject the Null Hypothesis')
```

#### Manual Calculation for reference using Formula

num = (x1_bar-x2_bar)

```python
# Step 1: Decide the test to perform. 
# 2 sample t test

# Step 2: Hypothesis Construction. 
# H0 : mu1 = mu2
# H1 : mu1 != mu2

# Step 3 : Decide the Direction of the test  
# two tail test

course_completed = df[df['test preparation course']=='completed']['total score']
course_not_completed = df[df['test preparation course']=='none']['total score']

import statistics
x1_bar = np.mean(course_completed) 
S1 = statistics.stdev(course_completed)
n1 = len(course_completed)
x2_bar = np.mean(course_not_completed)
S2 = statistics.stdev(course_not_completed)
n2 = len(course_not_completed)

num = (x1_bar-x2_bar) 
sp = np.sqrt( ((  (n1-1)*S1**2)  +((n2-1)*S2**2 ))   / (n1+n2-2)     )
den = np.sqrt(sp**2*(1/n1+1/n2))
test_stat=num/den
print('test_stat',test_stat)
cv = stats.t.isf(0.05/2,df=31) #n1+n2-2
print('CV',cv)
p_val = stats.t.sf(abs(test_stat),df=31)*2
print('p_val',p_val)
print('sig_lev',0.05)
```

```python
# since p_values is not less than sig_level we fail to reject H0.
```

The two samples have standard deviation of 4.1 g/dl for male donors and 3.5 g/dl for female donor . Can it be said the population means of concentrations of the elements are the same for men and women? Use  α = 0.01.

```python
#H0 : mu1 = mu2
#H1 : mu1 != mu2
# This is two tail test
x1_bar =13 
x2_bar = 15
sigma_1 = 4.1
sigma_2 = 3.5
n1 = 160
n2 = 180
test_stat = (x1_bar-x2_bar)/np.sqrt((sigma_1**2/n1)+(sigma_2**2/n2))
print('test_stat',test_stat)
cv = stats.norm.isf(0.01/2)
print('Critical Value',cv)
p_value = stats.norm.sf(abs(test_stat))*2
print('p_value',p_value)
print('Sig_level',0.01)
print(p_value<0.01)
#since p_value is less than sig_lev we reject H0, concluding that mu1!=mu2
```

```python
import pandas as pd
df = pd.read_csv('salary_data_demo.csv')
df.head(2)
```

#### Parametric Condition check: 

We perform parametric test if the data passess 2 assumptions: 
1. Data is Normal ( we use shapiro test to check this assumption)
2. Data has equal variacne ( we use levene test to check this assumption)

##### Pearson Correlation Test (Numeric vs Numeric)

```python
# Hypothesis Formation: 
# Null Hypothesis : Salary and Exp are Independent
# Alternate Hypothesis : Salary and Exp are Dependent 

# Performing test 
stats.pearsonr(df['Salary'],df['Exp'])

# Conclusion
#Since p_val 4.176240265838662e-12 is less than sig_level 0.05 we reject Null Hypothesis cocluding there is a relation among Salary and Exp 
```

##### 2 sample independent t test (Num vs Cat(2 Sub groups))

##### Normality Condition Check 

```python
#Null : Skew = 0 Data is Normal 
#Alter : Skew != 0   Data is not normal
male_salary  = df[df['Gender']=='Male']['Salary']
female_salary  = df[df['Gender']=='Female']['Salary']
print(stats.shapiro(male_salary))
print(stats.shapiro(female_salary))

# Since p_value 0.98 & 0.97 is not less than 0.05  we fail reject the null and conclude that the data is normal.
```

##### Equal Variance 

```python
# Null : Var is Equal 
# Alter : Var is not equal
male_salary  = df[df['Gender']=='Male']['Salary']
female_salary  = df[df['Gender']=='Female']['Salary']
stats.levene(male_salary,female_salary)
# Since p_value 0.93 is not less than 0.05  we fail reject the null and conclude that the data has equal variance.
```

```python
# Gender vs Salary

# Hypothesis Formation: 
# Null Hypothesis : Salary and Gender are Independent
# Alternate Hypothesis : Salary and Gender are Dependent 

# Performing test 

male_salary = df[df['Gender']=='Male']['Salary']
female_salary = df[df['Gender']=='Female']['Salary']
stats.ttest_ind(male_salary,female_salary)

# Conclusion
#Since p_val 0.9 is not less than sig_level 0.05 we fail to reject Null Hypothesis cocluding there is no relation among Salary and Gender
```

##### ANOVA (Num vs Cat(>2 Sub groups))

```python
# Gender vs Salary

# Hypothesis Formation: 
# Null Hypothesis : Salary and Education are Independent
# Alternate Hypothesis : Salary and Education are Dependent 

# Performing test 

PG_salary = df[df['Education']=='PG']['Salary']
UG_salary = df[df['Education']=='UG']['Salary']
salary_12th = df[df['Education']=='12th']['Salary']
stats.f_oneway(PG_salary,UG_salary,salary_12th)

# Conclusion
#Since p_val 6.612074692440832e-17 is not less than sig_level 0.05 we reject Null Hypothesis cocluding there is a relation among Salary and Education
```

##### Post-hoc (Tukey Test)

 Post Hoc Analysis will help us in understanding where exactly the sample means are different 


```python
import statsmodels.stats.multicomp as mc
post_hoc = mc.MultiComparison(data=df['Salary'],groups=df['Education'])
tukey = post_hoc.tukeyhsd()
tukey.summary()
```

```python
# If reject = True then the mean of those groups are different. 

# If rekect = False then the mean of those groups are not different.
```

### Non Parametric Test

##### CHI-SQUARE test (CAT vs CAT)

```python
# Gender vs Education
# Null : Gender & Education are Independent  /  Gender has no significance on Education

# Alter: Gender & Education are Dependent / Gender has significance on Education.
```

```python
pd.crosstab(df['Gender'],df['Education'])
```

```python
stats.chi2_contingency(pd.crosstab(df['Gender'],df['Education']))
```

```python
# Since p Value is not less than significance level we fail to reject the null hypothesis concluding that gender has no impact on education
```

##### Spearman Correlation test

```python
# Hypothesis Formation: 
# Null Hypothesis : Salary and Exp are Independent
# Alternate Hypothesis : Salary and Exp are Dependent 

# Performing test 
stats.spearmanr(df['Salary'],df['Exp'])

# Conclusion
#Since p_val 1.980034868083735e-10 is less than sig_level 0.05 we reject Null Hypothesis cocluding there is a relation among Salary and Exp 
```

##### Manwhiteny-U test

```python
# Gender vs Salary

# Hypothesis Formation: 
# Null Hypothesis : Salary and Gender are Independent
# Alternate Hypothesis : Salary and Gender are Dependent 

# Performing test 

male_salary = df[df['Gender']=='Male']['Salary']
female_salary = df[df['Gender']=='Female']['Salary']
stats.mannwhitneyu(male_salary,female_salary)

# Conclusion
#Since p_val 0.88 is not less than sig_level 0.05 we fail to reject Null Hypothesis cocluding there is no relation among Salary and Gender
```

##### Kruskal-Wallis test

```python
# Gender vs Salary

# Hypothesis Formation: 
# Null Hypothesis : Salary and Education are Independent
# Alternate Hypothesis : Salary and Education are Dependent 

# Performing test 

PG_salary = df[df['Education']=='PG']['Salary']
UG_salary = df[df['Education']=='UG']['Salary']
salary_12th = df[df['Education']=='12th']['Salary']
stats.kruskal(PG_salary,UG_salary,salary_12th)

# Conclusion
#Since p_val 8.720635031850622e-13 is not less than sig_level 0.05 we reject Null Hypothesis cocluding there is a relation among Salary and Education
```
