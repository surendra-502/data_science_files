### 3.What is Population and Sample?
The population is a complete set of elements having some common characteristics selected for a research study.<br>
A sample is a set of randomly chosen observations from a population. Samples are drawn from the population when a dataset is large since it is challenging to study the whole population. We draw a sample when the item under study is feared to exhaust. For instance,<br> 
<ul>
    <li>the study of the durability of an electric bulb </li>
        <li>the study of the effectiveness of new medicine</li>
</ul> 

#### Sampling Techniques 

#### Sampling without Replacement

In sampling without replacement, once an item is selected, it is not returned to the population before the next selection.
This means each item can be chosen only once.

Example:

Using the same 5 balls (A, B, C, D, E):

If you pick one ball and keep it aside (do not put it back), and then pick the next, that’s sampling without replacement.

```python
data = [1, 2, 3, 4, 5, 6]
n = 5 # Sample size

without_replacement = random.sample(data, n)

# Output
print("Population:", data)
print("Sample size:", n)
print("Without Replacement:", without_replacement)
```

**Note: We can draw defined number of Samples from this method**

#### Sampling with Replacement

In sampling with replacement, after selecting an item from the population, it is returned back before the next selection.
This means the same item can be chosen more than once.

Example:

Imagine you have 5 colored balls labeled A, B, C, D, E.

If you pick one ball, record it, and then put it back into the bag before picking the next — that’s sampling with replacement.

**Note: We can draw infinite number of Samples from this method**

```python

data = [1, 2, 3, 4, 5, 6]
n = 5 # Sample size

# Sampling WITH Replacement
with_replacement = random.choices(data, k=n)

# Output
print("Population:", data)
print("Sample size:", n)
print("Without Replacement:", with_replacement)
```

### 5.Distribution of the data

<ul>
    <li>A distribution in statistics shows how data or values are spread out or how frequently each value occurs in a dataset.
    <li>Think of it as a way of describing the pattern of your data.</li>
    <li>On plotting the histogram or a frequency curve for a variable, we are  actually looking at how the data is distributed over its range</li> 


### Types of Distributions:

There are many types, but the most commonly studied are:

Binomial Distribution

Normal Distribution

Standard Normal Distribution

#### Binomial Distribution: 

A Binomial Distribution is used when we perform an experiment that has only two possible outcomes:



#### Properties of Binomial Distribution

There are a fixed number of trials (n).

Each trial has only two possible outcomes (Success or Failure).

The probability of success (p) is the same in each trial.

All trials are independent (the result of one doesn’t affect another).

#### Normal Distribution

A Normal Distribution is a continuous probability distribution where data values are symmetrically distributed around the mean.
It means that most observations are close to the average (mean), its also called as Gaussian Distribution.?

#### Properties of Normal Distribution

The Distribution Follows a Bell Shape Curve. 

The Mean = Median = Mode will be coinciding. 

The data will be symmetric, left side data will be same as right side data.

#### Standard Normal Distribution

A Standard Normal Distribution is a special case of the normal distribution.

It has:  Mean = 0,  Standard Deviation = 1

We convert a normal variable into a standard normal variable using the Z-score.

#### Properties of Standard Normal Distribution

Mean (μ) = 0

Standard Deviation (σ) = 1

Total Area Under the Curve = 1

Empirical Rule Applies (68–95–99.7 rule)

```python
# Generate 1000 random values from a normal distribution
# loc/mean = 50, scale/standard deviation = 10
np.random.seed(42)  # for reproducibility
data = np.random.normal(loc=50, scale=10, size=1000)
data = pd.DataFrame(data,columns=['Price'])
data.describe()
```

```python
sns.histplot(data['Price'])
```

```python
# z_score method to convert Normal Distribution to Standard Normal Distribution
```

```python
mu = data['Price'].mean()
sigma = data['Price'].std()
x = data['Price']
```

```python
z_score = (x-mu)/sigma
```

```python
data['Standardized_Price'] = z_score
```

```python
np.round(data['Standardized_Price'].mean())
```

```python
np.round(data['Standardized_Price'].std())
```

```python
sns.histplot(data['Standardized_Price'])
```

##### About 68% of data will be between -1 to +1.

##### About 95% of data will be between -2 to +2.

##### About 95% of data will be between -3 to +3.
