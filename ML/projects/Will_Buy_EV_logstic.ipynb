```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/train.csv")
df.head(2)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()  #checking null values.no null values in the data so, the data is clean
```

```python
df.duplicated().sum()#checking duplicates no duplicates
```

```python
df.describe(exclude=np.number)
```

```python
df['Will_Buy_EV'].value_counts(normalize=True)
```

```python
x=df.drop(columns=['Will_Buy_EV'])
y=df['Will_Buy_EV']
```

```python
x.head(1)
```

```python
x['Daily_Commute_km']=x['Daily_Commute_km'].fillna(x['Daily_Commute_km'].median())
x['Number_of_Cars_Owned']=x['Number_of_Cars_Owned'].fillna(x['Number_of_Cars_Owned'].median())

x['Charging_Stations_Near_Home']=x['Charging_Stations_Near_Home'].fillna(x['Charging_Stations_Near_Home'].median())

x['Charging_Stations_Near_Work']=x['Charging_Stations_Near_Work'].fillna(x['Charging_Stations_Near_Work'].median())
```

```python
x.isnull().sum()
```

```python
x['Environmental_Concern_Level']=x['Environmental_Concern_Level'].fillna(x['Environmental_Concern_Level'].median())
x['Range_Anxiety_Level']=x['Range_Anxiety_Level'].fillna(x['Range_Anxiety_Level'].mode()[0])
x['Gender']=x['Gender'].fillna(x['Gender'].mode()[0])
x['City_Type']=x['City_Type'].fillna(x['City_Type'].mode()[0])
x['Current_Car_Type']=x['Current_Car_Type'].fillna(x['Current_Car_Type'].mode()[0])
x['Home_Charging_Possible']=x['Home_Charging_Possible'].fillna(x['Home_Charging_Possible'].mode()[0])
x['Subsidy_Available']=x['Subsidy_Available'].fillna(x['Subsidy_Available'].mode()[0])
```

```python
x.isnull().sum()
```

```python
y.head(1)
```

```python
y = y.fillna(y.mode()[0])
```

```python
#encoding categorical variables

cat_columns=x.select_dtypes(exclude=np.number)
cat_columns.head(2)
```

```python
num_data=x.select_dtypes(include=np.number)
num_data.head(2)
```

```python
#lets check their unique values. Because type of encoding depends on no of categories
for col in cat_columns:
  print('\n',col)
  print(cat_columns[col].value_counts())
```

```python
y = y.map({'Yes':1,'No':0})
```

```python
encoded_data=pd.get_dummies(cat_columns[['Gender','City_Type','Current_Car_Type']],drop_first=True,dtype=int)
```

```python
cat_columns[['Range_Anxiety_Level']].isnull().sum()
```

```python
from sklearn.preprocessing import OrdinalEncoder
categories=['Low','Medium','High']
OE=OrdinalEncoder(categories=[categories])
cat_columns['Range_Anxiety_Level']=OE.fit_transform(cat_columns[['Range_Anxiety_Level']])
```

```python
cat_columns['Home_Charging_Possible']=cat_columns['Home_Charging_Possible'].map({'Yes':1,'No':0})
cat_columns['Subsidy_Available']=cat_columns['Subsidy_Available'].map({'Yes':1,'No':0})
```

```python
final_encoded_data=pd.concat([cat_columns,encoded_data],axis=1)
```

```python
final_encoded_data.head(2)
```

```python

final_encoded_data.drop(columns=['Gender','City_Type','Current_Car_Type'],inplace=True)
```

```python
final_encoded_data.head(1)
```

```python
#now need to combine a numerical variables and encoded_variables
final_df=pd.concat([final_encoded_data,num_data],axis=1)
```

```python
final_df.head(2)
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(final_df,y,train_size=0.8,random_state=100)
```

```python
print(xtrain.shape)
print(xtest.shape)
print(ytrain.shape)
print(ytest.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
xtrain[['Age','Annual_Income_USD','Daily_Commute_km']]=sc.fit_transform(xtrain[['Age','Annual_Income_USD','Daily_Commute_km']])
xtest[['Age','Annual_Income_USD','Daily_Commute_km']]=sc.transform(xtest[['Age','Annual_Income_USD','Daily_Commute_km']])
```

```python
xtrain.head(2)
```

```python

xtrain.drop(columns=['id'],inplace=True)
xtest.drop(columns=['id'],inplace=True)
```

```python
xtrain.head(2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
lr=LogisticRegression(max_iter=1000)
lr.fit(xtrain,ytrain)
ypred=lr.predict(xtest)
```

```python
print(classification_report(ytest,ypred))
```

```python
y.head(2)
```

```python
y_prob=lr.predict_proba(xtest)[:,1]
```

```python
from sklearn.metrics import roc_auc_score
roc_auc_score(ytest,y_prob)
```
