```python
import pandas as pd
import numpy as np 
```

```python
df = pd.read_excel('test.xlsx')
df
```

```python
x = df[['x']]
y = df['y'] 
from sklearn.linear_model import LinearRegression # import the model from library 
LR = LinearRegression()  # Initiate the model 
LR.fit(x,y)# Fit the Model

print('b0 intercept:',LR.intercept_)
print('b1 slope:',LR.coef_) 
```
